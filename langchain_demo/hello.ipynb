```python
import pandas as pd
import numpy as np

# 模拟一个工业场景的数据：生成 10 条生产线数据
data = {
    'machine_id': ['M01', 'M02', 'M03', 'M04', 'M01', 'M02', 'M05', 'M01', 'M03', 'M02'],
    'temperature': [55.2, 60.1, 45.8, 88.5, 54.3, 61.2, 22.0, 56.1, 46.5, 92.1],
    'status': ['Normal', 'Normal', 'Normal', 'Overheat', 'Normal', 'Normal', 'Off', 'Normal', 'Normal', 'Overheat']
}

# 把字典变成 DataFrame (类似 Excel 表格)
df = pd.DataFrame(data)
# 这里为了演示你可以直接把 data 丢进去：
df = pd.DataFrame(data)

# 运行完这格，你看不到输出，但数据已经进内存了
print("数据加载完成！")
```
